### <font color = '#DEB887'> Pandas - Data Cleaning

```python
# import data.csv
df = pd.read_csv("data.csv")

df
```

```python
# Check the existence of null values 
df.isnull()
```

```python
# Using 'sum' funtion (counts the number of missing variables  in each column)
df.isnull().sum()
```

```python
# Using 'any' funtion (counts the number of columns with missing variables)
df.isnull().any().sum()
```

```python
# using 'isna' function
df.isna().sum()
```

```python
# Delete null values using dropna 
df.dropna(inplace = True)

df
```

#### NOTE: It is important to know the distribution of data before cleaning

```python
# import data.csv
df = pd.read_csv("data.csv")

df
```

```python
# Replace null values by the mean (for numeric variables (when data is symmetric))
for col in df.columns[:2]:
    df[col].fillna(df[col].mean(), inplace = True)
    
df
```

```python
# import data.csv
df = pd.read_csv("data.csv")

df
```

```python
# Replace null values by the median (for numerical variables (when data is skewed))
for col in df.columns[:2]:
    df[col].fillna(df[col].median(), inplace = True)
    
df
```

```python
# import data.csv
df = pd.read_csv("data.csv")

df
```

```python
# Replace null values by the mode (for categorical and numerical variables)
for col in df.columns[:2]:
    df[col].fillna(df[col].mode()[0], inplace = True)
    
df
```

```python
# import data.csv
df = pd.read_csv("data.csv")

df
```
